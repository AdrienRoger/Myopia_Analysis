# 1. Introduction: Predictive Modeling and Insights from Previous Analyses

In this section, we build on the findings from the initial exploratory and statistical analyses ([notebook link provided in the project files](1.Data_exploration.ipynb)) to guide our predictive modeling strategy. These investigations have revealed the **multifactorial nature of childhood myopia** and highlighted the features most relevant for robust prediction.

### Key Findings and Data Insights from Initial Analyses

**1. Family History is a Significant but Non-Deterministic Factor:**

* **PARENTSMY (At least one myopic parent):** Strongly associated with myopia risk, but not an absolute determinant. Many children with myopic parents remain non-myopic, and conversely, some without parental myopia do develop it, underlining the **complex interplay of genetic and environmental factors**.

**2. SPHEQ Dominates as a Discriminant Variable:**

* **SPHEQ (Spherical equivalent refraction):** The single most predictive feature for myopia. Statistically significant (very low p-value) and consistently top-ranked in feature importance analyses across all models. Lower SPHEQ values strongly indicate myopia.
* **ACD (Anterior chamber depth):** Shows a notable difference between myopic and non-myopic children, suggesting a structural ocular component to risk.

**3. Physical Activity Shows Limited but Consistent Association:**

* **SPORTHR (Hours in sports/outdoors):** Myopic children tend to report slightly fewer hours of sport. While statistically significant, the practical magnitude of this effect appears modest.

**4. Features with No Meaningful Association:**

* **GENDER:** No statistically significant association with myopia (p > 0.05) and consistently low predictive importance.
* **SCREENHR, CLOSEHR, AGE, LT, VCD, STUDYYEAR:** Most of these features do not demonstrate a direct significant link with myopia individually. However, some (e.g. VCD, LT) are involved in structural correlations and may still contribute within multivariate or interaction-based models.

**5. Inter-feature Correlations Reveal Latent Structures:**

* **AL-VCD:** Strong positive correlation indicating a shared underlying ocular structural influence.
* **LT correlations:** LT is correlated with both ACD and VCD, suggesting anatomical dependencies.
* **SCREENHR-CLOSEHR:** Moderate correlation, reflecting overlapping behavioral patterns.

---

### UMAP + OPTICS Clustering Insights

Although dimensionality reduction (UMAP) and clustering (OPTICS) were explored to identify potential **latent subgroups or risk profiles**, no clustering solution achieved both a **high silhouette score** and **robust clustering with minimal outliers**. This suggests the dataset’s structure is **continuous and overlapping**, not naturally forming discrete groups in the reduced space.

---

### Implications for Predictive Modeling

> Building upon these insights, our predictive modeling work will:
>
> * **Prioritize sensitivity** to detect at-risk myopic children, addressing class imbalance via SMOTE, class weighting, and threshold tuning.
> * **Focus on the strongest predictors** (SPHEQ, PARENTSMY, ACD) while engineering new interactions to capture multifactorial patterns.
> * **Consider inter-feature correlations** to avoid redundancy and improve model efficiency.
> * **Test non-linear models** to capture complex interactions not evident in univariate tests.
> * **Continuously monitor model fairness** across subgroups to ensure equitable clinical applicability.

The next modeling steps will integrate these findings to build **transparent, robust, and clinically actionable predictive models for childhood myopia risk.**

# 3. Predicting Modeling

## 3.1. Feature engineering

````python
df_train['MYOPIC'] = df_train['MYOPIC'].astype(int)
df_train['SPORTHR*2']=df_train['SPORTHR']*2
df_train['SPHEQ*5']=df_train['SPHEQ']*5

df_test['MYOPIC'] = df_test['MYOPIC'].astype(int)
df_test['SPORTHR*2']=df_test['SPORTHR']*2
df_test['SPHEQ*5']=df_test['SPHEQ']*5
````

````python
X_train = df_train.drop(['MYOPIC', 'SPORTHR*2', 'SPHEQ*5'], axis=1)
X_test = df_test.drop(['MYOPIC', 'SPORTHR*2', 'SPHEQ*5'], axis=1)
````

````python
x_test = df_test.drop(['MYOPIC', 'SPORTHR', 'SPHEQ'], axis=1)
x_test['PARENTSMY'] = x_test['PARENTSMY'].astype(int)
y_test = df_test['MYOPIC'].astype(int)

x_train = df_train.drop(['MYOPIC', 'SPORTHR', 'SPHEQ'], axis=1)
x_train['PARENTSMY'] = x_train['PARENTSMY'].astype(int)
y_train = df_train['MYOPIC'].astype(int)
````

#### Commentary

Based on the initial exploratory analyses and the identified modeling challenges, we introduce advanced feature engineering steps here:

- Higher-order terms (e.g., SPORTHR$*2$, SPHEQ$*5$) are created to capture possible non-linear relationships, especially in cases where standard linear effects showed limited discrimination power or potential ambiguity around clinical cutoffs (as seen for SPHEQ).
- The focus on parental myopia as a binary feature aligns with prior findings that family risk is important but not absolute, and may interact with other predictors.
- SPORTHR and SPHEQ are removed after generating their non-linear transform to reduce collinearity while retaining potentially meaningful non-linear effects.
- The dataset is prepared for modeling by separating independent variables (`x`) from the target (`y`), reflecting the need for model-ready input highlighted in the synthesis table.

These steps directly address earlier observations: non-linearity and feature interaction may help resolve ambiguity near decision boundaries, while variable selection mitigates redundancy. Subsequent modeling will further balance predictive performance and interpretability, with a focus on minority class (myopic) recall.

## 3.2. Selection of variables

### 3.2.1. VIF Data and Interaction

````python
vif_data = pd.DataFrame()
vif_data['feature'] = x_train.columns
vif_data['VIF'] = [variance_inflation_factor(x_train.values, i) for i in range(x_train.shape[1])]
(vif_data)
````

#### Commentary

The VIF (Variance Inflation Factor) analysis confirms the presence of **strong multicollinearity** among certain ocular biometric features—specifically **AL, ACD, LT**, and **VCD**. This finding is fully consistent with our earlier correlation analyses and pairplots, which highlighted high interdependencies within these variables.

Such multicollinearity can lead to **unstable model coefficients** and reduced interpretability, as these features carry **overlapping information** about ocular structure. Addressing these redundancies is therefore critical to ensure both model robustness and clarity in identifying risk drivers.

To mitigate this issue, our strategy will be to **group or reduce** these correlated features. In particular, because **ACD** has demonstrated strong individual predictive power in previous analyses, we will retain it as a standalone variable while **aggregating AL, VCD, and LT** to avoid inflating the model with redundant biometric information.

This approach preserves the essential structural signals for myopia prediction while enhancing model stability and interpretability.

### 3.2.2. PCA

````python
x_train_pca = x_train.copy()
features_bio1 = ['AL', 'VCD', 'LT']
pca = PCA(n_components=2)
bio_train_pca1 = pca.fit_transform(x_train_pca[features_bio1])

#x_pca = x.drop(columns=features_bio)
x_train_pca[['PC1_BIO_1', 'PC2_BIO_1']] = bio_train_pca1
````

````python
x_test_pca = x_test.copy()
bio_test_pca1 = pca.fit_transform(x_test_pca[features_bio1])

x_test_pca[['PC1_BIO_1', 'PC2_BIO_1']] = bio_test_pca1
````

````python
x_train_pca = x_train_pca.drop(vif_data['feature'][vif_data['VIF']>10].tolist(), axis=1)
x_train_pca = pd.concat([x_train_pca, x_train['ACD']], axis=1)

vif_data2 = pd.DataFrame()
vif_data2['feature'] = x_train_pca.columns
vif_data2['VIF'] = [variance_inflation_factor(x_train_pca.values, i) for i in range(x_train_pca.shape[1])]
(vif_data2)
````

````python
x_test_pca = x_test_pca.drop(vif_data['feature'][vif_data['VIF']>10].tolist(), axis=1)
x_test_pca = pd.concat([x_test_pca, x_test['ACD']], axis=1)
````

````python
if "PARENTSMY" not in x_train_pca.columns.tolist():
    x_train_pca = pd.concat([x_train_pca, pd.DataFrame(x_train['PARENTSMY'], columns=['PARENTSMY'])], axis=1)
if "PARENTSMY" not in x_test_pca.columns.tolist():
    x_test_pca = pd.concat([x_test_pca, pd.DataFrame(x_test['PARENTSMY'], columns=['PARENTSMY'])], axis=1)
````

### 3.2.3. Combinations

````python
# train
features = x_train_pca.columns.tolist()
interactions = []
for comb in combinations(features, 2):
    name = comb[0]+":"+comb[1]
    df_train[name] = x_train_pca[comb[0]] * x_train_pca[comb[1]]
    interactions.append(name)
X2_train = pd.concat([x_train_pca, df_train[interactions]], axis=1)

# test
features = x_test_pca.columns.tolist()
interactions = []
for comb in combinations(features, 2):
    name = comb[0]+":"+comb[1]
    df_test[name] = x_test_pca[comb[0]] * x_test_pca[comb[1]]
    interactions.append(name)
X2_test = pd.concat([x_test_pca, df_test[interactions]], axis=1)
````

````python
# train
features = x_train_pca.columns.tolist()
new_feats_train = {}

for comb in combinations(features, 3):
    name = f"{comb[0]}:{comb[1]}:{comb[2]}"
    new_feats_train[name] = x_train_pca[comb[0]] * x_train_pca[comb[1]] * x_train_pca[comb[2]]

interactions_df_train = pd.DataFrame(new_feats_train)
interactions_df_train.index = X2_train.index

X2_train = pd.concat([X2_train, interactions_df_train], axis=1)

# test
features = x_test_pca.columns.tolist()
new_feats_test = {}

for comb in combinations(features, 3):
    name = f"{comb[0]}:{comb[1]}:{comb[2]}"
    new_feats_test[name] = x_test_pca[comb[0]] * x_test_pca[comb[1]] * x_test_pca[comb[2]]

interactions_df_test = pd.DataFrame(new_feats_test)
interactions_df_test.index = X2_test.index

X2_test = pd.concat([X2_test, interactions_df_test], axis=1)
````

#### Commentary

To further enhance the model's ability to capture complex, non-linear relationships between features, all possible **2-way, 3-way, and 4-way** feature interactions are systematically generated. This is especially relevant after the application of PCA, as principal components can encode shared structure, but may still benefit from multiplicative interaction terms that represent higher-order dependencies or risk factor synergies.

### 3.2.3. LASSO and RFE - Automatic Selection Variables

````python
n_features_to_select = 30
````

````python
scaler = StandardScaler()
X_scaled_train = scaler.fit_transform(X2_train)
X_scaled_test = scaler.transform(X2_test)
````

````python
sel = RFE(LogisticRegression(solver='liblinear'), n_features_to_select=n_features_to_select)
sel = sel.fit(X_scaled_train, y_train)
coefs = sel.estimator_.coef_[0]
selected_features = X2_train.columns[sel.support_]

# Tri des features selon leur importance absolue
variables_selected_RFE = pd.Series(np.abs(coefs), index=selected_features).sort_values(ascending=False)
print("Features triées par importance (valeur absolue du coefficient):")
variables_selected_RFE = variables_selected_RFE.index.tolist()
print(variables_selected_RFE)
````

#### Commentary

In this step, we perform feature selection using Recursive Feature Elimination (RFE), aiming to identify the most relevant predictors for myopia classification. Logistic regression is used as the base estimator to generate a ranking of features, with an initial selection of 30 variables. This setup ensures consistency and interpretability, especially as the exact number of features will later be optimized in conjunction with model hyperparameters.

Instead of fixing the number of features upfront, we treat it as a tunable component. RFE provides an ordered list of variables based on their contribution to the logistic model, and we evaluate the downstream performance of various classifiers (logistic regression, random forest, and gradient boosting) using subsets of top-ranked features. This approach enables us to balance model simplicity with predictive accuracy.

Because we now have a clear ranking of the most important features, we can proceed to model-specific optimization—tuning each classifier not only in terms of hyperparameters but also by selecting the optimal number of features to include.

## 3.4. Model Optimisation 

#### Optimization Approach:

1. **Quick Exploratory Phase:**

   * Initial hyperparameter tuning for each model based on the **Average Precision (AP)** metric, to avoid overfitting and guide further optimization.
   * `RandomizedSearchCV` is used for each model on a preselected set of features.
   * **Goal:** Identify a **relevant search space** for each hyperparameter.

2. **Fine Optimization Phase:**

   * Re-optimization of the number of features using models pre-tuned with the initial hyperparameters.
   * `Optuna` is employed with **bounds defined from the randomized search results**.
   * **Goal:** Find optimal hyperparameters with better **exploration vs. exploitation** trade-off.
   * Final search for the **optimal number of features** to retain.

### 3.4.1 Metrics

#### functions

````python
def eval_model(model, X_train, y_train, X_test, y_test, seuil=0.5, name='model', cv=5):
    model.fit(X_train, y_train)
    model = CalibratedClassifierCV(model, method='sigmoid', cv=5)
    model.fit(X_train, y_train)
    y_pred_proba = model.predict_proba(X_test)[:, 1]
    y_pred_label = (y_pred_proba > seuil).astype(int)
    print(f"\n===== {name} =====")
    print("Accuracy:", accuracy_score(y_test, y_pred_label))
    print("AUC:", roc_auc_score(y_test, y_pred_proba))
    print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_label))
    print(classification_report(y_test, y_pred_label))
    
    ## Analysis
    test_results = X_test.copy()
    test_results['y_true'] = y_test
    test_results['y_pred'] = y_pred_label
    test_results['proba_pred'] = y_pred_proba
    return test_results, model

def eval_by_group(X, y_true, y_pred, group_col):
    groups = X[group_col].unique()
    for grp in groups:
        idx = X[group_col] == grp
        print(f"\n--- {group_col} = {grp} ---")
        print(classification_report(y_true[idx], y_pred[idx]))

def plot_precision_recall_threshold(model, X_train, y_train, X_test, y_test, cv=5):    
    # Entraînement
    model.fit(X_train, y_train)
    y_proba = model.predict_proba(X_test)[:, 1]

    # Courbes précision/rappel en fonction du seuil
    precisions, recalls, ths = precision_recall_curve(y_test, y_proba)
    plt.plot(ths, precisions[:-1], label='Precision')
    plt.plot(ths, recalls[:-1], label='Recall')
    plt.xlabel("Threshold")
    plt.ylabel("Score")
    plt.title("Precision / Recall by Threshold")
    plt.legend()
    plt.grid(True)
    plt.show()


def plot_calibration_curve(model, X_train, y_train, X_test, y_test, method='isotonic', cv=5, label=None):
    # Calibration du modèle
    cal_model = CalibratedClassifierCV(model, method=method, cv=cv)
    cal_model.fit(X_train, y_train)

    # Prédictions calibrées
    y_cal_proba = cal_model.predict_proba(X_test)[:, 1]
    prob_true, prob_pred = calibration_curve(y_test, y_cal_proba, n_bins=10)

    # Affichage de la courbe de calibration
    plt.plot(prob_pred, prob_true, marker='o', label=label or f'{model.__class__.__name__} calibré')
    plt.plot([0, 1], [0, 1], 'k--', label='Perfect calibration')
    plt.xlabel("Predicted probability")
    plt.ylabel("Observed frequency")
    plt.title("Calibration curve")
    plt.legend()
    plt.grid(True)
    plt.show()

    return cal_model  # Pour pouvoir utiliser le modèle calibré ensuite

def plot_precision_recall_curve(model, X_test, y_test, label=None):
    # Prédiction des probabilités
    y_proba = model.predict_proba(X_test)[:, 1]
    
    # Calcul des courbes
    precision, recall, _ = precision_recall_curve(y_test, y_proba)
    ap = average_precision_score(y_test, y_proba)

    # Affichage
    plt.plot(recall, precision, label=f"{label or model.__class__.__name__} (AP={ap:.2f})")
    plt.xlabel("Recall")
    plt.ylabel("Precision")
    plt.title("Precision-Recall Curve")
    plt.legend()
    plt.grid(True)
    plt.show()
````

#### a. Logistic Regression

````python
from sklearn.calibration import CalibratedClassifierCV, calibration_curve
from sklearn.metrics import precision_recall_curve, average_precision_score
import matplotlib.pyplot as plt

# Logistic Regression
print("="*30, 'Logistic Regression', "="*30)

lr = LogisticRegression(
    C= 0.043579687785260314, 
    penalty= 'l2', 
    max_iter= 1619,
    random_state = 5
)

test_results_lr, model_lr = eval_model(
    lr, X_train_lr, y_resampled_lr, X_test_lr, y_test, name='Logistic Regression', seuil=0.7
)

# Feature importance
plt.figure(figsize=(8, 5))
coefs = pd.Series(model_lr.estimator.coef_[0], index=X_train_lr.columns)

coefs_sorted = coefs.abs().sort_values()

explainer = shap.Explainer(model_lr.estimator, X_train_lr) 
shap_values = explainer(X_test_lr)

# Affichage (beeswarm ou bar : importance feature, etc.)
shap.summary_plot(shap_values, X_test_lr, plot_type="bar")
shap.summary_plot(shap_values, X_test_lr)  # beeswarm

plot_precision_recall_threshold(lr, X_train_lr, y_resampled_lr, X_test_lr, y_test)
cal_rf = plot_calibration_curve(lr, X_train_lr, y_resampled_lr, X_test_lr, y_test, label='Logistic Regression')
plot_precision_recall_curve(lr, X_test_lr, y_test, label="Logistic Regression")
````
